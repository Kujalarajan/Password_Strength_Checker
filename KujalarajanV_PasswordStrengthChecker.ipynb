## Aim

## Method
The `zxcvbn` estimator checks predictable patterns and returns a score from 0 (very weak) to 4 (very strong). The displayed attack-time estimate is illustrative and depends on the selected attack scenario.


```python
# Install the project dependency if it is not already installed.
%pip install -r requirements.txt
```

```python
from getpass import getpass
from zxcvbn import zxcvbn

def assess_password(password: str) -> dict:
    """Assess a password locally and return a compact result."""
    if not isinstance(password, str):
        raise TypeError("Password must be text.")
    if not password:
        return {"score": 0, "label": "Very Weak", "guesses": 1,
                "crack_time": "instant",
                "feedback": ["Enter a password to receive an assessment."]}

    result = zxcvbn(password)
    labels = ["Very Weak", "Weak", "Fair", "Strong", "Very Strong"]
    feedback = []
    warning = result.get("feedback", {}).get("warning", "")
    suggestions = result.get("feedback", {}).get("suggestions", [])
    if warning:
        feedback.append(warning)
    feedback.extend(suggestions)
    if len(password) < 12:
        feedback.append("Consider a longer passphrase (12–16+ characters).")
    if not feedback:
        feedback.append("Use this password for only one account.")

    return {
        "score": result["score"],
        "label": labels[result["score"]],
        "guesses": result.get("guesses", 0),
        "crack_time": result.get("crack_times_display", {}).get(
            "offline_slow_hashing_1e4_per_second", "not available"),
        "feedback": feedback,
    }

def display_result(result: dict) -> None:
    print("\\n--- Password Strength Report ---")
    print(f"Strength: {result['label']} ({result['score']}/4)")
    print(f"Estimated guesses: {result['guesses']:,}")
    print(f"Illustrative offline attack estimate: {result['crack_time']}")
    print("\\nSuggestions:")
    for suggestion in result["feedback"]:
        print("-", suggestion)
    print("\\nPrivacy: this notebook does not save or transmit the password.")
```

## Test with sample inputs
Use made-up examples only. Do not paste a password used for a real account into a shared computer or demonstration notebook.


```python
sample_passwords = ["123456", "welcome123", "River!Cloud7-Mango"]
for sample in sample_passwords:
    print("\\nSample password:", "*" * len(sample))
    display_result(assess_password(sample))
```

## Interactive assessment
Run this cell to assess a password. Input is hidden where the notebook environment supports `getpass`. Use a made-up test password for classroom demonstrations.


```python
password = getpass("Enter a password to assess: ")
display_result(assess_password(password))
del password  # Remove the variable reference after the assessment.
```
